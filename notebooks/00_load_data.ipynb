```python
from pathlib import Path
from datasets import load_dataset
from tqdm import tqdm
```

```python
# definimos las rutas 
PROJECT_ROOT = Path().resolve().parent
RAW_PATH = PROJECT_ROOT / "data" / "raw"
RAW_PATH.mkdir(parents=True, exist_ok=True)

IMAGES_PATH = RAW_PATH 
IMAGES_PATH.mkdir(exist_ok=True)

RAW_PATH
```

```python
# descargamos el cache desde hugging face
dataset = load_dataset("nielsr/CelebA-faces", split="train")
```

```python
# guardamos sólo una parte de las imágenes (MAX_IMAGES)
MAX_IMAGES = 15000

for i, example in enumerate(tqdm(dataset)):
    example["image"].save(IMAGES_PATH / f"{i:06d}.jpg")
    
    if i + 1 >= MAX_IMAGES:
        break

print(f"Se guardaron {MAX_IMAGES} imágenes en {IMAGES_PATH}")
```

```python
# verificamos que se hayan guardado las imágenes que indicamos en la carpeta raw
total_images = len(list(IMAGES_PATH.glob("*.jpg")))
print(f"Total imágenes en raw: {total_images}")
```
